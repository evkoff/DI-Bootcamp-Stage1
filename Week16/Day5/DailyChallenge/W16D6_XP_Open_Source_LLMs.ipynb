```python

# Store the comparison table in Markdown format
comparison_table = """| Openness level | What's open? | Impact on retraining/modifying |
| --- | --- | --- |
| Fully Open | Architecture, pretrained weights, source code | Can inspect, modify, retrain, and fine-tune the entire model |
| Weights Released | Pretrained weights and architecture | Can fine-tune and run inference, but cannot reproduce the original training process |
| Architecture Only | Model architecture | Can study or reimplement the design, but must train a new model from scratch |
"""

print(comparison_table)

from IPython.display import Markdown
Markdown(comparison_table)
```

```python

# TODO: 3?5 sentence comparative paragraph of the three openness levels
comparative_paragraph = """
Open-source LLMs differ in how much of the model is publicly available. Fully Open models provide the architecture, pretrained weights, and source code, making them the best choice for research and complete customization. Models with Weights Released allow fine-tuning and inference but do not expose the full training pipeline. Architecture Only models reveal the model design but require training from scratch because pretrained weights are unavailable. Choosing the right openness level depends on how much control and flexibility your project requires.
"""

print(comparative_paragraph)
```
