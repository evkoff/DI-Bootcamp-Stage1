## Exercise 2: Multi-Part Prompt for Quiz Generation
**Scenario**: Create an educational prompt for 7th graders about volcanic eruptions.

### Tasks
- Ask for: 2-bullet summary + 3 MCQs (1 correct, 2 distractors).
- Require friendly tone, simple vocab (11?13 y/o), and Google Slides-ready output.
- Explain 3 ways your prompt improves on: `Make a quiz for kids about this article.`

```python
# TODO: write the full multi-part prompt
quiz_prompt = """
Act as an educational content creator for 7th-grade students.

Using only the provided article about volcanic eruptions:

1. Summarize the article in exactly 2 simple bullet points.
2. Create 3 age-appropriate multiple-choice questions based on the article.
3. For each question, provide 3 answer options: 1 correct answer and 2 distractors.
4. Use a friendly tone and simple vocabulary suitable for 11–13-year-olds.
5. Format the output so it is easy to copy into Google Slides.
6. Clearly label the correct answer for each question.

Do not add information that is not supported by the article.
"""
quiz_prompt
```

```python
# TODO: explain three improvements over the vague prompt
improvements = [
    'It clearly defines the required output: a 2-bullet summary and 3 multiple-choice questions.',
    'It specifies the target age, friendly tone, and simple vocabulary, making the content appropriate for 11–13-year-olds.',
    'It controls the answer format by requiring 1 correct answer and 2 distractors per question and a Google Slides-ready structure.'
]
improvements
```
