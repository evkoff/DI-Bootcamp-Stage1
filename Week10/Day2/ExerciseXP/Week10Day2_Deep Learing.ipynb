**What is the key difference between traditional machine learning and deep learning?**

Traditional machine learning usually requires manual feature engineering, where humans decide which features are important for the model. Deep learning automatically learns useful features directly from raw data using multiple neural network layers. Example: To classify images of cats, a traditional machine learning model requires manually designed features such as ear shape or fur color, while a deep learning model learns these features automatically from thousands of images.

**How do artificial neural networks (ANN) mimic the human brain?**

Artificial neural networks are inspired by the structure of the human brain. They consist of interconnected artificial neurons that receive inputs, apply weights and activation functions, and pass information to other neurons to learn patterns from data.

** Why does deep learning perform better on large datasets compared to traditional machine learning?**

Deep learning performs better on large datasets because neural networks can learn complex patterns and hierarchical features automatically. With more data, they improve their ability to generalize, while traditional machine learning models often rely on manually designed features.

**What are some challenges of deep learning, and how can they be addressed?**

Some challenges include the need for large datasets, high computational power, long training times, and the risk of overfitting. These challenges can be addressed by using GPUs, data augmentation, regularization techniques such as dropout, and transfer learning.

**What is feature engineering, and why is it not needed in deep learning?**

Feature engineering is the process of manually selecting or creating useful input features for a machine learning model. In deep learning, the neural network automatically learns the most relevant features from raw data during training.

**What role do hidden layers play in a deep learning model?**

Hidden layers transform the input data into increasingly abstract representations. Each hidden layer learns more complex patterns, allowing the network to solve complicated tasks such as image recognition and language understanding.

**In an artificial neural network (ANN), what is the function of an activation function?**

An activation function introduces non-linearity into the neural network, allowing it to learn complex relationships between inputs and outputs. Without activation functions, a neural network would behave like a simple linear model regardless of the number of layers. Example: Without an activation function, a neural network would not be able to distinguish between handwritten digits such as "3" and "8", because it would behave like a simple linear model.

## 🌟 Exercise 3: Building a Simple Neural Network with TensorFlow or Keras

**As stated in the instructions**  
Build a classifier for MNIST with:
- One input layer
- One hidden layer with 128 neurons and ReLU activation
- One output layer with 10 neurons and softmax activation

Steps:
1. Import TensorFlow and needed modules from `tensorflow.keras`
2. Load the MNIST dataset
3. Normalize data by dividing by 255.0
4. One-hot encode the labels
5. Build a sequential model with Flatten, Dense(128, ReLU), Dense(10, softmax)
6. Compile with optimizer `adam`, loss `categorical_crossentropy`, metrics `accuracy`
7. Train and evaluate on the test set

**Guidance**  
You will execute prefilled boilerplate for imports and dataset loading. You will complete the one-hot encoding and the model specification, compile, train, and evaluate.

**Learning point**  
Normalization stabilizes optimization. One-hot encoding matches softmax outputs to target vectors. ReLU accelerates training by avoiding saturation, while softmax maps logits to a probability simplex.

![image.png](https://github.com/user-attachments/assets/61737b80-69cb-41a1-aea5-76092cd2a373)


```python
# PREFILLED: just execute
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.utils import to_categorical #for One-Hot Encoding.
from tensorflow.keras.datasets import mnist # mnist - a dataset from Tensorflow library, contains 70000 images of handwritten digits

print("TensorFlow:", tf.__version__)

# Load data
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalize to [0,1]
x_train = x_train.astype("float32") / 255.0 #normalizattion
x_test  = x_test.astype("float32") / 255.0

# Flatten will be added in the model; keep images as 28x28 for now
print("Train shape:", x_train.shape, "Test shape:", x_test.shape)
```

```python
# To-Do: one-hot encode labels y_train and y_test using to_categorical
# Hints: use to_categorical
y_train_oh = to_categorical(y_train)
y_test_oh = to_categorical(y_test)

print("Labels one-hot:", y_train_oh.shape, y_test_oh.shape)
```

```python
# To-Do:  Build a sequential model with Flatten, Dense(128, ReLU), Dense(10, softmax)
model = models.Sequential([
    layers.Flatten(input_shape=(28, 28)), #result is a vector of length = 784
    layers.Dense(128, activation="relu"), # Dense(128, ReLU) - hidden layer with 128 neurons and ReLU activation, where 128 is a hyperparameter, RelU - activation function
    layers.Dense(10, activation="softmax"), # Dense(10, softmax) - output layer with 10 neurons and softmax activation
])

model.summary()
```

```python
# To-Do: compile the model with adam (oprimizer) (to updates weights), categorical_crossentropy (to measure loss), and accuracy (metric)
model.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])
```

```python
# To-Do: train then evaluate
history = model.fit(
    x_train,
    y_train_oh,
    epochs=5,
    batch_size=32
)
test_loss, test_acc = model.evaluate(
    x_test,
    y_test_oh
)
print({"test_loss": float(test_loss), "test_acc": float(test_acc)})
```

## 🌟 Exercise 5 (optional): Implementing Forward and Backward Propagation in Python

**As stated in the enoncé**  
You will code a simple neural network that performs forward propagation and backpropagation for a regression problem predicting exam scores from study hours.

**Guidance**  
Run the prefilled code. Then modify the learning rate or initial weights and observe changes in the loss and updates.
Provide a short written explanation.

**Learning point**  
Gradient descent updates parameters in the direction of steepest loss reduction.
The learning rate controls step size. Too small slows learning. Too large risks divergence.


```python
# PREFILLED: just execute
import numpy as np

# Initialize input data (features)
x = np.array([4, 80])  # 4 hours studied, previous test score: 80

# Initialize weights and bias
w = np.array([0.6, 0.3])  # Initial weights
b = 10  # Initial bias

# Forward Propagation
def forward_propagation(x, w, b):
    z = np.dot(x, w) + b  # Weighted sum
    return z  # Linear activation (No ReLU here, it's a regression task)

# Compute prediction
y_pred = forward_propagation(x, w, b)
y_true = 85  # Actual exam score

# Compute Loss (Mean Squared Error)
loss = 0.5 * (y_true - y_pred) ** 2

# Compute Gradients (Backward Propagation)
grad_w = -(y_true - y_pred) * x  # Partial derivatives with respect to weights
grad_b = -(y_true - y_pred)  # Partial derivative with respect to bias
# minus says Go to the direction opposite to the loss (error) growth direction.

# Update Weights and Bias
learning_rate = 0.01
w_new = w - learning_rate * grad_w
b_new = b - learning_rate * grad_b

# Print Results
print("Initial Prediction:", y_pred)
print("Loss:", loss)
print("Updated Weights:", w_new)
print("Updated Bias:", b_new)
```

**To-Do:** Explain in full sentences why updating weights using gradient descent reduces the error. Then modify the learning rate or initial weights and describe how this affects learning.


Gradient descent reduces the error by updating the weights and bias in the direction that minimizes the loss function. The gradients show how each parameter affects the error, and the learning rate controls the size of each update. After updating the parameters, the model usually makes more accurate predictions and the loss decreases. I increased the learning rate from 0.01 to 0.1. The weights changed much more in one step. This can speed up learning, but if the learning rate is too large, the model may overshoot the optimal values and become unstable.
