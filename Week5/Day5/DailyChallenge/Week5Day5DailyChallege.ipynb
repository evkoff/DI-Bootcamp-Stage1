```python
#Data Preparation: Download and explore the US Superstore data. Perform basic data cleaning and preprocessing.
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://github.com/devtlv/Datasets-DA-Bootcamp-2-/raw/refs/heads/main/Week%205%20-%20Data%20Processing/W5D5%20-%20Mini-project%20-%20bis/US%20Superstore%20data.xls"
df = pd.read_excel(url)
df.head()
```

```python
df.info()
df.shape
```

```python
df[df.isnull().any(axis=1)]
```

```python
df[df.duplicated()]
```

```python
df = df.dropna()

df = df.drop_duplicates()
df.shape
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df.head()
```

```python
#Sales Trends Over Years
df["Year"] = df["Order Date"].dt.year
yearly_sales = df.groupby("Year")["Sales"].sum()
yearly_sales
```

```python
#interactive line chart
#Data Visualization with Matplotlib: Create an interactive line chart to show sales trends over the years.
plt.figure(figsize=(10,6)) #creates 'canvas' for the chat and defines its size
plt.plot(
    yearly_sales.index,
    yearly_sales.values,
    marker="o"
    )
plt.xlabel("Year")
plt.ylabel("Sales")
plt.title("Sales Trends Over Years")
plt.grid(True)
plt.xticks(yearly_sales.index)
plt.show()
```

```python
#Sales Distribution by Country
df["Country"].value_counts()
```

```python
#Build an interactive map to visualize sales distribution by country.
state_sales = df.groupby("State")["Sales"].sum()
plt.figure(figsize=(14,6))
plt.bar(
    state_sales.index,
    state_sales.values
    )
plt.xticks(rotation=90)
plt.title("Sales Distribution by State")
plt.show()
```

```python
#Top 10 products by sales (Seaborn)
product_sales = df.groupby("Product Name")["Sales"].sum() #the result will be series, not df (so it will containt index and values)
product_sales = product_sales.sort_values(ascending=False)
top10_products = product_sales.head(10)
top10_products
```

```python
type(product_sales)
```

```python
#Data Visualization with Seaborn: Use Seaborn to generate a bar chart showing top 10 products by sales.
plt.figure(figsize=(12,6))
sns.barplot(
    x=top10_products.values,
    y=top10_products.index
)
plt.title("Top 10 Products by Sales")
plt.show()
```

```python
#Create a scatter plot to analyze the relationship between profit and discount.
# Что такое scatterplot - Каждая точка = одна строка dataset.
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=df,
    x="Discount",
    y="Profit"
)
plt.title("Profit vs Discount")
plt.show()
```

Key Insights
Sales increased over the years, showing business growth.
California and New York contribute significantly to total sales.
A small number of products generate a large proportion of revenue.
Higher discounts are often associated with lower profitability.
Seaborn provides cleaner statistical visualizations, while Matplotlib offers greater customization flexibility.
