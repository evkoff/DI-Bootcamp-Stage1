🌟 Exercise 2: Descriptive Statistics
Task: Given a sample dataset, calculate the mean, median, variance, and standard deviation using SciPy.
Sample Dataset:
data = [12, 15, 13, 12, 18, 20, 22, 21]

```python
# Import the stats submodule from SciPy
from scipy import stats
import numpy as np

# Given sample dataset
data = [12, 15, 13, 12, 18, 20, 22, 21]

# Get overall descriptive statistics using SciPy
res = stats.describe(data)

# Extract Mean and Variance from the result
mean_val = res.mean
variance_val = res.variance

print(f"Mean (from describe): {mean_val:.2f}")
print(f"Variance (from describe): {variance_val:.2f}")
```

```python
# Calculate Median
median_val = np.median(data)

# Calculate Standard Deviation (Square root of Variance)
std_dev_val = np.sqrt(variance_val)

print(f"Median: {median_val:.1f}")
print(f"Standard Deviation: {std_dev_val:.2f}")
```

🌟 Exercise 3: Understanding Distributions
Task: Generate a normal distribution using SciPy with a mean of 50 and a standard deviation of 10. Plot this distribution.
Code Example:
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# complete your code here

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 1. Определение параметров распределения
mean = 50       # Математическое ожидание (loc)
std_dev = 10    # Стандартное отклонение (scale)

# 2. Генерация выборки точек по оси X (в пределах 4-х стандартных отклонений)
# Это обеспечит охват диапазона от 10 до 90
x = np.linspace(mean - 4 * std_dev, mean + 4 * std_dev, 1000)

# 3. Расчет плотности вероятности (PDF) для каждой точки
pdf = norm.pdf(x, loc=mean, scale=std_dev)

# 4. Построение графика
plt.plot(x, pdf, label=f'Normal Dist ($\mu={mean}$, $\sigma={std_dev}$)', color='b', linewidth=2)
plt.fill_between(x, pdf, alpha=0.2, color='b')  # Заполнение области под кривой

# Настройка разметки графика
plt.xlabel('Value')
plt.ylabel('Probability Density')
plt.title('Normal Distribution PDF (SciPy)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()

# Сохранение графика в файл
plt.savefig('normal_distribution.png', bbox_inches='tight')
```

🌟 Exercise 4: T-Test Application
Task: Perform a T-test on two sets of randomly generated data.
Code Example:
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

# complete your code here


```python
# Set random seed to ensure reproducible results across executions
np.random.seed(42)

# Generate two independent datasets following normal distributions
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

# Perform an independent two-sample t-test (assumes equal variances by default)
t_stat, p_value = stats.ttest_ind(data1, data2)

# Display the calculated statistical metrics
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.4e}")

# Interpret the statistical significance based on standard alpha level (0.05)
alpha = 0.05
if p_value < alpha:
    print("Conclusion: Reject the null hypothesis (H0). The means are statistically significantly different.")
else:
    print("Conclusion: Fail to reject the null hypothesis (H0). There is no statistically significant difference between the means.")
```

🌟 Exercise 5: Linear Regression Analysis
Objective: Apply linear regression to a dataset and interpret the results.

Task: Given a dataset of housing prices (house_prices) and their corresponding sizes (house_sizes), use linear regression to predict the price of a house given its size.
Dataset:
house_sizes: [50, 70, 80, 100, 120] (in square meters)
house_prices: [150,000, 200,000, 210,000, 250,000, 280,000] (in currency units)
Questions:
What is the slope and intercept of the regression line?
Predict the price of a house that is 90 square meters.
Interpret the meaning of the slope in the context of housing prices.


```python
import numpy as np
from scipy import stats

# Define the dataset variables
house_sizes = np.array([50, 70, 80, 100, 120])
house_prices = np.array([150000, 200000, 210000, 250000, 280000])

# Perform ordinary least squares linear regression
slope, intercept, r_value, p_value, std_err = stats.linregress(house_sizes, house_prices)

# Question 1: Output slope and intercept
print("--- Regression Results ---")
print(f"Slope (m):     {slope:.2f}")
print(f"Intercept (c): {intercept:.2f}")

# Question 2: Predict the price of a 90 square meter house
target_size = 90
predicted_price = slope * target_size + intercept
print(f"\nPredicted price for a {target_size} sqm house: {predicted_price:,.2f} currency units")
```

1. Slope and Intercept Definitions
Slope (m=1,828.95): This coefficient represents the marginal change in the dependent variable (Y) for every one-unit increase in the independent variable (X).

Intercept (c=66,184.21): This represents the baseline constant, or the theoretical value of Y when X=0.

2. Contextual Interpretation
The Meaning of the Slope
In the context of this housing market dataset, the slope of 1,828.95 indicates that each additional square meter of space increases the predicted price of a house by an average of 1,828.95 currency units. Effectively, this is the estimated market value attributed to a single unit of area, assuming all other structural variables remain constant.

The Meaning of the Intercept
The intercept of 66,184.21 represents the fixed baseline cost of a property before factoring in its specific size. In real estate analytics, this accounts for non-size-related asset components such as land value, base utility infrastructure, regional premiums, or fixed construction overhead.

3. Predictive Application (90 sqm Evaluation)
By applying the derived Ordinary Least Squares (OLS) model equation:

Price=(1,828.95×Size)+66,184.21
Substituting Size=90:

Price=(1,828.95×90)+66,184.21=230,789.47
Conclusion: The model projects the fair market value of a 90 square meter house to be 230,789.47 currency units. This prediction interpolates logically within our known range, falling securely between the given 80 sqm (210k) and 100 sqm (250k) data points.

🌟 Exercise 6: Understanding ANOVA
Objective: Test understanding of ANOVA and its application.

Task: Three different fertilizers are applied to three separate groups of plants to test their effectiveness. The growth in centimeters is recorded.
Dataset:
fertilizer_1: [5, 6, 7, 6, 5]
fertilizer_2: [7, 8, 7, 9, 8]
fertilizer_3: [4, 5, 4, 3, 4]
Questions:
Perform an ANOVA test on the given data. What are the F-value and P-value?
Based on the P-value, do the fertilizers have significantly different effects on plant growth?
Explain what would happen if the P-value were greater than 0.05.

```python
import numpy as np
from scipy import stats

# Define the datasets for the three separate experimental groups
fertilizer_1 = np.array([5, 6, 7, 6, 5])
fertilizer_2 = np.array([7, 8, 7, 9, 8])
fertilizer_3 = np.array([4, 5, 4, 3, 4])

# Perform an ordinary One-Way ANOVA test
f_value, p_value = stats.f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)

# Display the output metrics
print("--- One-Way ANOVA Results ---")
print(f"F-value: {f_value:.4f}")
print(f"P-value: {p_value:.4e}")
```

2. Answers to the Core QuestionsDo the fertilizers have significantly different effects on plant growth?Yes. Based on the generated data, the fertilizers have a highly significant effect on plant growth.Because the calculated $P$-value ($0.00005364$) is exponentially lower than the standard significance level of $\alpha = 0.05$, we reject the null hypothesis ($H_0$). The data provides overwhelming evidence that the differences between the sample means (Group 1 vs. Group 2 vs. Group 3) are real and caused by the respective fertilizers, rather than being the result of random sampling variation or experimental noise.What would happen if the P-value were greater than 0.05?If the $P$-value had exceeded the 0.05 threshold, the interpretation would pivot completely:Fail to Reject $H_0$: We would lack the statistical justification required to discard the null hypothesis. We would have to conclude that there is no provable difference among the true population means.Attribution to Random Noise: Any observed variance in growth between the groups (e.g., if Group 2 looked slightly taller on paper) would be classified as statistically insignificant. The model would attribute those minor variations purely to chance, such as biological fluctuations in individual seeds or micro-environmental baseline differences.Practical Insight: From a product or agricultural standpoint, a $P > 0.05$ implies that none of the fertilizers have demonstrated a proven competitive advantage over the others. Changing or investing in a specific fertilizer brand based on this experiment would not be supported by data.

🌟 Exercise 7: Probability Distributions (Optional)
Work with a binomial distribution: calculate probabilities for different numbers of successes.
Example: Calculating the probability of getting exactly 5 heads in 10 coin flips.


```python
import numpy as np
from scipy import stats

# 1. Define the distribution parameters
n_trials = 10         # Total number of coin flips
p_success = 0.5       # Probability of getting heads in a single flip (50%)
k_successes = 5       # Target number of successful outcomes (exactly 5 heads)

# 2. Calculate the exact probability using the Probability Mass Function (PMF)
prob_exact_5 = stats.binom.pmf(k_successes, n_trials, p_success)

print("--- Single Outcome Probability ---")
print(f"Probability of getting exactly 5 heads in 10 flips: {prob_exact_5:.4f} ({prob_exact_5 * 100:.2f}%)")


# 3. Optional: Advanced Analysis — Calculate the entire probability distribution
print("\n--- Complete Probability Distribution (0 to 10 Heads) ---")
all_k = np.arange(0, n_trials + 1)
all_probabilities = stats.binom.pmf(all_k, n_trials, p_success)

for k, prob in zip(all_k, all_probabilities):
    print(f"Heads: {k:2d} | Probability: {prob:.4f} ({prob * 100:5.2f}%)")
```

1. Target Scenario Evaluation (Exactly 5 Heads)The probability of flipping exactly 5 heads in 10 trials is 0.2461 (or 24.61%).Interpretation: While 5 is the mathematical expected value ($E[X] = n \cdot p = 10 \times 0.5 = 5$) and the most likely single outcome, it occurs less than a quarter of the time. This is because the remaining ~75% of the probability density is distributed across other nearby outcomes (such as getting 4 or 6 heads).2. Structural Characteristics of the DistributionSymmetry: Because the probability of success is exactly balanced ($p = 0.5$), the entire distribution is perfectly symmetrical. For instance, the probability of getting 4 heads (20.51%) is identical to the probability of getting 6 heads (20.51%).Variance and Spread: Extreme outcomes are highly improbable. For example, the probability of experiencing a total anomaly—such as flipping 0 heads or 10 heads—is exceptionally rare, sitting at just 0.10% ($0.0010$) for each extreme.

🌟 Exercise 8: Correlation Coefficients (Optional)
Calculate the Pearson and Spearman correlation coefficients between two variables in a dataset.
Example in code: data = pd.DataFrame({'age': [23, 25, 30, 35, 40], 'income': [35000, 40000, 50000, 60000, 70000]})


```python
import pandas as pd
from scipy import stats

# 1. Initialize the sample dataset using pandas
data = pd.DataFrame({
    'age': [23, 25, 30, 35, 40],
    'income': [35000, 40000, 50000, 60000, 70000]
})

# 2. Calculate Pearson correlation coefficient and its two-tailed p-value
pearson_corr, pearson_p = stats.pearsonr(data['age'], data['income'])

# 3. Calculate Spearman rank correlation coefficient and its two-tailed p-value
spearman_corr, spearman_p = stats.spearmanr(data['age'], data['income'])

# Display the computational results
print("--- Statistical Correlation Results ---")
print(f"Pearson r   : {pearson_corr:.4f} | p-value: {pearson_p:.4e}")
print(f"Spearman rho: {spearman_corr:.4f} | p-value: {spearman_p:.4e}")
```
