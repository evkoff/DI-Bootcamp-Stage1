# Load Dataset

```python
# Import the dataset loader from LangChain
from langchain_community.document_loaders import HuggingFaceDatasetLoader

# Specify the Hugging Face dataset path
dataset_path = "databricks/databricks-dolly-15k"

# Specify which column will be treated as document content
page_content_column = "context"

# Create the dataset loader
loader = HuggingFaceDatasetLoader(
    path=dataset_path,
    page_content_column=page_content_column
)

# Load the dataset as LangChain documents
data = loader.load()

# Display the first two documents
print(data[:2])
```

# Split Documents

```python
# Import the text splitter
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Create a text splitter with chunk overlap
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150
)

# Split documents into smaller chunks
docs = text_splitter.split_documents(data)

# Display the first chunk
print(docs[0])
```

```python
# Compare the number of documents before and after splitting
print(f"Original documents: {len(data)}")
print(f"Chunks after splitting: {len(docs)}")
```

```python
# Find documents that were actually split into multiple chunks
long_docs = [doc for doc in data if len(doc.page_content) > 1000]

print(f"Documents longer than 1000 characters: {len(long_docs)}")
```

# Generate Embeddings

```python
# Install the Hugging Face integration for LangChain
!pip install -q langchain-huggingface

# Import the embeddings model
from langchain_huggingface import HuggingFaceEmbeddings

# Specify the embedding model
model_name = "sentence-transformers/all-MiniLM-L6-v2"

# Configure the device
model_kwargs = {
    "device": "cpu"
}

# Configure embedding generation
encode_kwargs = {
    "normalize_embeddings": False
}

# Create the embeddings model
embeddings = HuggingFaceEmbeddings(
    model_name=model_name,
    model_kwargs=model_kwargs,
    encode_kwargs=encode_kwargs
)

# Test embedding generation
text = "This is a test document."

embedding = embeddings.embed_query(text)

print(embedding[:10])
print(len(embedding))
```

# FAISS Index (Build Vector Store)

```python
# Import the FAISS vector store
from langchain_community.vectorstores import FAISS

# Create a FAISS vector store from the document chunks
db = FAISS.from_documents(
    docs,
    embeddings
)

print(type(db))
```

```python
# Display the number of indexed documents
print(db.index.ntotal)
```

# Create Retriever

```python
# Create a retriever from the FAISS vector store
retriever = db.as_retriever(
    search_kwargs={"k": 4}
)

# Display the retriever object
print(retriever)
```

```python
# Import Hugging Face pipeline
from transformers import pipeline

# Import LangChain wrapper
from langchain_huggingface import HuggingFacePipeline

# Specify the LLM
model_id = "google/flan-t5-base"

# Create a text generation pipeline
pipe = pipeline(
    task="text2text-generation",
    model=model_id,
    max_new_tokens=100
)

# Wrap the pipeline for LangChain
llm = HuggingFacePipeline(pipeline=pipe)
```

# Retriever


```python
import langchain

print(langchain.__version__)
print(langchain.__file__)
```

```python
import langchain_classic

print(langchain_classic.__version__)
```

```python
# Import modern LangChain components
from langchain_core.prompts import ChatPromptTemplate
from langchain_classic.chains.combine_documents import create_stuff_documents_chain
from langchain_classic.chains.retrieval import create_retrieval_chain

# Create the retriever
retriever = db.as_retriever(search_kwargs={"k": 4}) # give me 4 most relevant documents

print(type(retriever))
```

# Prompt


```python
# Create the prompt template
prompt = ChatPromptTemplate.from_template(
    """Answer the question using only the provided context.

If the answer is not in the context, say "I don't know."

Context:
{context}

Question:
{input}
"""
)

print(prompt)
```

# Document Chain

```python
# Create the document chain
document_chain = create_stuff_documents_chain(
    llm=llm,
    prompt=prompt
)

print(type(document_chain))
```

# Retrieval Chain

```python
# Create the retrieval chain
qa = create_retrieval_chain(
    retriever,
    document_chain
)

print(type(qa))
```

# Ask Questions

```python
# Ask a question
response = qa.invoke(
    {
        "input": "When did Virgin Australia start operating?"
    }
)

print(response)
```
