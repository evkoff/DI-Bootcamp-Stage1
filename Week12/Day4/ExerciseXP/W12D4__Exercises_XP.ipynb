## 🌟 Exercise 2: LLM Architecture and Application Scenarios
For each, describe (a) core architectural differences, (b) a real application, (c) why it fits.


### BERT

- **Architecture:** Bidirectional encoder-only Transformer trained with Masked Language Modeling (MLM) (and historically Next Sentence Prediction).
- **Application:** Sentiment analysis / Classification / NER / Q&A.
- **Why:** BERT considers both left and right context, making it highly effective for text understanding and classification tasks.

### GPT

- **Architecture:** Decoder-only Transformer trained with Causal Language Modeling (autoregressive next-token prediction).
- **Application:** Chatbots and text generation.
- **Why:** GPT generates text one token at a time, making it ideal for producing coherent and natural-sounding responses.

### T5

- **Architecture:** Encoder-decoder Transformer using a text-to-text framework.
- **Application:** Machine translation / summarization.
- **Why:** T5 is designed to map one text sequence to another, making it well suited for sequence-to-sequence tasks such as translation and summarization.

## 🌟 Exercise 3: Benefits and Ethics of Pre-training
  ### Benefits

1. **Improved generalization:** Pre-trained models learn general language patterns, allowing them to perform well on new tasks and unseen data.

2. **Less labeled data:** Since the model already has language knowledge, it requires much less labeled data for fine-tuning.

3. **Faster fine-tuning:** Fine-tuning starts from a strong pre-trained model, making training faster than starting from scratch.

4. **Transfer learning:** Knowledge learned during pre-training can be transferred to many different NLP tasks, improving performance.

5. **Robustness:** Pre-trained models are generally more robust to variations in wording and can better handle unfamiliar inputs.

### Ethical concerns

Pre-trained models may learn biases present in the training data, generate misinformation, expose private information, or be misused for harmful purposes such as spam or fake content generation.

### Mitigations

These risks can be reduced through careful data curation, differential privacy techniques, safety filters, reinforcement learning from human feedback (RLHF), and regular model audits.


## 🌟 Exercise 4: Transformer Architecture Deep Dive

## Self-Attention

Self-attention allows each token to attend to every other token in the input sequence. Each token is transformed into **Query (Q)**, **Key (K)**, and **Value (V)** vectors. Attention scores are computed by comparing Queries with Keys, then normalized using the **softmax** function. The resulting weights are used to compute a weighted sum of the Value vectors, allowing each token to gather relevant contextual information.

## Multi-Head Attention

Multi-head attention uses several attention heads that learn different relationships between words in parallel. Each head projects the input into a different subspace and focuses on different aspects of the sentence, such as syntax, semantics, or long-range dependencies. This enables the model to capture richer contextual information than a single attention head.

### Example

**Sentence:** *The scientist thanked the assistant because she solved the problem.*

- One attention head may focus on the relationship between **"she"** and **"assistant"**, resolving the pronoun reference.
- Another attention head may focus on the relationship between **"solved"** and **"problem"**, identifying the action and its object.

## 🌟 Exercise 6: Softmax Temperature: The Randomness Regulator

## 1. Temperature Scenarios

- **T = 0.2:** The model produces more deterministic, focused, and predictable outputs. It is less creative but more consistent.
- **T = 1.5:** The model generates more diverse and creative responses but may produce less coherent or less accurate text.
- **T = 1.0:** The model balances creativity and coherence, providing varied yet generally reliable responses.

## 2. Application Design

### Bedtime stories

For bedtime stories, I would use a **temperature around 0.9–1.1**. This provides a good balance between creativity and coherence, making the stories imaginative while still easy for children to follow.

### Financial report summaries

For financial summaries, I would use a **low temperature (around 0.2–0.4)** to produce accurate, consistent, and reliable outputs while minimizing unnecessary creativity.

## 3. Temperature and Bias
- TODO: How does temperature influence bias surfacing or dampening? Give a realistic example.

Increasing the temperature can make biased or unusual outputs more likely because the model samples from a wider range of possible tokens. Lower temperatures usually produce more consistent and predictable responses but do not eliminate bias learned during training. For example, at a high temperature, a model generating job recommendations might produce more stereotypical or less reliable suggestions, while a lower temperature would tend to produce more consistent and factual responses.


### (Optional) Quick Generation Demo Across Temperatures
Note: Requires `transformers` and model download; skip if offline.


```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
import torch

model_name = "gpt2"  # TODO: choose a causal LM
prompt = "A robot and a cat decided to explore Mars together."  # TODO: write your own prompt

pipe = pipeline(
    "text-generation",
    model=AutoModelForCausalLM.from_pretrained(model_name), # from_pretrained() means - download existing model (take ready-made weights)
    tokenizer=AutoTokenizer.from_pretrained(model_name),
    device=0 if torch.cuda.is_available() else -1,
)

for T in [0.2, 1.0, 1.5]:
    out = pipe(prompt, max_new_tokens=40, temperature=T, do_sample=True) # do_sample=True means to allow random choice
    print("--- Temperature:", T)
    print(out[0]["generated_text"])  # Inspect how style changes with T; out[0] because pipeline always returns list of dictionaries
```
