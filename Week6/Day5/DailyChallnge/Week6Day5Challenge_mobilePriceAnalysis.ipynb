```python
# Import pandas library
import pandas as pd

# Load dataset from csv file
df = pd.read_csv('train.csv')

# Show first 5 rows
df.head()
```

```python
# Show information about dataset
df.info()
```

```python
#Conduct basic descriptive statistics for an overview of the dataset.
df.describe()
```

```python
df.isnull().sum()
```

```python
#Transform categorical data into numerical format using suitable methods
# Check unique values in categorical columns

categorical_columns = [
    'blue',
    'dual_sim',
    'four_g',
    'three_g',
    'touch_screen',
    'wifi'
]

for column in categorical_columns:
    print(column)
    print(df[column].unique())
    print()
```

```python
# Statistical Analysis with NumPy and SciPy

import numpy as np
from scipy import stats

# Calculate mean
mean_ram = np.mean(df['ram'])

# Calculate median
median_ram = np.median(df['ram'])

# Calculate mode
mode_ram = stats.mode(df['ram'])

# Print results
print("Mean RAM:", mean_ram)
print("Median RAM:", median_ram)
print("Mode RAM:", mode_ram)
```

```python
#Analysis of variability
# Calculate range
range_ram = np.max(df['ram']) - np.min(df['ram'])

# Calculate variance
variance_ram = np.var(df['ram'])

# Calculate standard deviation
std_ram = np.std(df['ram'])

# Print results
print("Range RAM:", range_ram)
print("Variance RAM:", variance_ram)
print("Standard Deviation RAM:", std_ram)
```

```python
#Evaluation of distribution shapes through skewness and kurtosis
# Calculate skewness
skewness_ram = stats.skew(df['ram'])

# Calculate kurtosis
kurtosis_ram = stats.kurtosis(df['ram'])

# Print results
print("Skewness RAM:", skewness_ram)
print("Kurtosis RAM:", kurtosis_ram)
```

```python
#Investigate feature-target correlations using SciPy
# Calculate correlation between RAM and price range

correlation_ram = stats.pearsonr(df['ram'], df['price_range'])

# Print result
print(correlation_ram) #correlation coefficient and p-value
```

```python
#Correlation Matrix. Analyze correlations between all features
# Calculate correlation matrix

correlation_matrix = df.corr()

# Show correlation matrix
correlation_matrix
```

```python
#RAM has the strongest correlation with mobile price classification, while features like WiFi and touch screen show minimal impact.
#Perform hypothesis testing

# Select RAM values for cheap phones
cheap_ram = df[df['price_range'] == 0]['ram']

# Select RAM values for expensive phones
expensive_ram = df[df['price_range'] == 3]['ram']

# Perform t-test
t_test_result = stats.ttest_ind(cheap_ram, expensive_ram)

# Print result
print(t_test_result)
```

```python
# Create a function to calculate descriptive statistics for one feature
def calculate_statistics(column):
    return {
        'mean': np.mean(column),
        'median': np.median(column),
        'mode': stats.mode(column, keepdims=True).mode[0],
        'range': np.max(column) - np.min(column),
        'variance': np.var(column),
        'std': np.std(column),
        'skewness': stats.skew(column),
        'kurtosis': stats.kurtosis(column)
    }




# Select all features except the target variable
features = df.drop(columns='price_range')

# Apply the function to every feature
statistics = features.apply(calculate_statistics)

# Convert the dictionaries into a DataFrame
statistics = pd.DataFrame(statistics.tolist(), index=features.columns)

# Display the results
statistics
```

```python
# Perform t-tests between the cheapest and most expensive price ranges

hypothesis_results = []

for feature in features.columns:
    cheap_group = df[df['price_range'] == 0][feature]
    expensive_group = df[df['price_range'] == 3][feature]

    t_stat, p_value = stats.ttest_ind(cheap_group, expensive_group)

    hypothesis_results.append({
        'feature': feature,
        't_statistic': t_stat,
        'p_value': p_value
    })

# Convert results into a DataFrame
hypothesis_results = pd.DataFrame(hypothesis_results)

# Add statistical significance
hypothesis_results['significant'] = hypothesis_results['p_value'] < 0.05

# Display results
hypothesis_results
```

```python
# Calculate Pearson correlation for every feature with the target

correlation_results = []

for feature in features.columns:
    correlation, p_value = stats.pearsonr(df[feature], df['price_range'])

    correlation_results.append({
        'feature': feature,
        'correlation': correlation,
        'p_value': p_value
    })

# Convert results into a DataFrame
correlation_results = pd.DataFrame(correlation_results)

# Sort by absolute correlation strength
correlation_results = correlation_results.sort_values(
    by='correlation',
    key=abs,
    ascending=False
)

# Display results
correlation_results
```

```python
# Perform ANOVA test across all four price ranges

anova_results = []

for feature in features.columns:
    groups = [
        df[df['price_range'] == price][feature]
        for price in range(4)
    ]

    f_stat, p_value = stats.f_oneway(*groups)

    anova_results.append({
        'feature': feature,
        'f_statistic': f_stat,
        'p_value': p_value
    })

# Convert results into a DataFrame
anova_results = pd.DataFrame(anova_results)

# Add statistical significance
anova_results['significant'] = anova_results['p_value'] < 0.05

# Sort by p-value
anova_results = anova_results.sort_values('p_value')

# Display results
anova_results
```

## Data Visualization with Matplotlib

### Histograms

```python
import matplotlib.pyplot as plt

# Select important numerical features
selected_features = ['ram', 'battery_power', 'px_width', 'px_height']

# Create histograms
for feature in selected_features:
    plt.figure(figsize=(7, 4))
    plt.hist(df[feature], bins=20)
    plt.title(f'Distribution of {feature}')
    plt.xlabel(feature)
    plt.ylabel('Frequency')
    plt.show()
```

## Insight Synthesis and Conclusion

The analysis shows that RAM is the strongest determinant of mobile phone price classification. It has a very strong positive Pearson correlation with price range (r = 0.917), and the difference in RAM between the cheapest and most expensive phones is statistically significant (p < 0.05). The box plot also shows a clear increase in RAM across the four price ranges.

Battery power, pixel width, and pixel height also show positive relationships with price range. ANOVA indicates statistically significant differences across price groups for RAM, battery power, pixel dimensions, mobile weight, internal memory, and number of cores.

In contrast, features such as WiFi, touch screen, Bluetooth, dual SIM, and clock speed show very weak correlations with price range and do not show statistically significant differences across the price groups in the hypothesis tests.

Overall, RAM appears to be the most important feature for distinguishing mobile phone price categories, while several other hardware specifications provide additional information. The visualizations and statistical tests consistently support these findings.
