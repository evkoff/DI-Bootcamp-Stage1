```python
# Import pandas library
import pandas as pd

# Load dataset from csv file
df = pd.read_csv('train.csv')

# Show first 5 rows
df.head()
```

```python
# Show information about dataset
df.info()
```

```python
#Conduct basic descriptive statistics for an overview of the dataset.
df.describe()
```

```python
df.isnull().sum()
```

```python
#Transform categorical data into numerical format using suitable methods
# Check unique values in categorical columns

categorical_columns = [
    'blue',
    'dual_sim',
    'four_g',
    'three_g',
    'touch_screen',
    'wifi'
]

for column in categorical_columns:
    print(column)
    print(df[column].unique())
    print()
```

```python
# Statistical Analysis with NumPy and SciPy

import numpy as np
from scipy import stats

# Calculate mean
mean_ram = np.mean(df['ram'])

# Calculate median
median_ram = np.median(df['ram'])

# Calculate mode
mode_ram = stats.mode(df['ram'])

# Print results
print("Mean RAM:", mean_ram)
print("Median RAM:", median_ram)
print("Mode RAM:", mode_ram)
```

```python
#Analysis of variability
# Calculate range
range_ram = np.max(df['ram']) - np.min(df['ram'])

# Calculate variance
variance_ram = np.var(df['ram'])

# Calculate standard deviation
std_ram = np.std(df['ram'])

# Print results
print("Range RAM:", range_ram)
print("Variance RAM:", variance_ram)
print("Standard Deviation RAM:", std_ram)
```

```python
#Evaluation of distribution shapes through skewness and kurtosis
# Calculate skewness
skewness_ram = stats.skew(df['ram'])

# Calculate kurtosis
kurtosis_ram = stats.kurtosis(df['ram'])

# Print results
print("Skewness RAM:", skewness_ram)
print("Kurtosis RAM:", kurtosis_ram)
```

```python
#Investigate feature-target correlations using SciPy
# Calculate correlation between RAM and price range

correlation_ram = stats.pearsonr(df['ram'], df['price_range'])

# Print result
print(correlation_ram) #correlation coefficient and p-value
```

```python
#Correlation Matrix. Analyze correlations between all features
# Calculate correlation matrix

correlation_matrix = df.corr()

# Show correlation matrix
correlation_matrix
```

```python
#RAM has the strongest correlation with mobile price classification, while features like WiFi and touch screen show minimal impact.
#Perform hypothesis testing

# Select RAM values for cheap phones
cheap_ram = df[df['price_range'] == 0]['ram']

# Select RAM values for expensive phones
expensive_ram = df[df['price_range'] == 3]['ram']

# Perform t-test
t_test_result = stats.ttest_ind(cheap_ram, expensive_ram)

# Print result
print(t_test_result)
```
