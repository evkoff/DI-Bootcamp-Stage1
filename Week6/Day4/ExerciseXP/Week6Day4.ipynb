🌟 Exercise 1 : Matrix Operations
Instructions
In this exercise, you’ll work with a 3x3 matrix. Here’s a brief explanation of the concepts:

Determinant: The determinant is a value that can be computed from the elements of a square matrix. It provides important information about the matrix, such as whether it has an inverse, and is used in various areas like linear algebra and calculus. To understand more about it you can watch this video.
Inverse of a Matrix: The inverse of a matrix is a matrix that, when multiplied with the original matrix, results in an identity matrix. Not all matrices have inverses. The inverse is crucial in solving systems of linear equations.

Create a 3x3 matrix and perform the following operations:

Calculate the determinant.
Find the inverse of the matrix.

```python
# Import the NumPy library to handle matrix operations
import numpy as np
# Create a 3x3 matrix using standard Python lists converted to a NumPy array
matrix = np.array([[4, 7, 2], [3, 6, 1], [2, 5, 1]])

# Display the created matrix to verify its structure
print("Original Matrix:")
print(matrix)
```

```python
# Calculate the determinant of the matrix using the linear algebra submodule
det = np.linalg.det(matrix)

# Display the calculated determinant value
print("\nDeterminant:")
print(det)
```

```python
# Compute the multiplicative inverse of the matrix
inverse_matrix = np.linalg.inv(matrix)

# Display the resulting inverse matrix
print("\nInverse Matrix:")
print(inverse_matrix)
```

```python
# Multiply the original matrix by its inverse to check the result
check_identity = matrix @ inverse_matrix

print("\nChecking the result (Should be Identity Matrix):")
print(check_identity)
```

```python
# Round the results to 10 decimal places to eliminate tiny floating-point errors
print(np.round(check_identity, decimals=10))
```

🌟 Exercise 6 : Basic Hypothesis Testing
Instructions
Create a sample dataset to test the effectiveness of a new training program on employee productivity:

```python
import numpy as np


# Set random seed for reproducibility
np.random.seed(42)

# Productivity scores of employees before the training program
# (Mean score is 50, standard deviation is 10)
productivity_before = np.random.normal(loc=50, scale=10, size=30)

# Productivity scores of the same employees after the training program
# (We add a random positive shift with a mean of 5 to simulate improvement)
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)
```

```python
# Calculate the individual productivity improvement for each employee
differences = productivity_after - productivity_before

# Calculate the mean (average) improvement across all employees
mean_improvement = np.mean(differences)

# Display the average improvement value
print(f"Average productivity increase: {mean_improvement:.2f}")
```

```python
# Calculate the standard deviation of the improvements
std_diff = np.std(differences, ddof=1)

# Calculate the number of samples (employees)
n = len(differences)

# Compute the Standard Error of the Mean (SEM)
standard_error = std_diff / np.sqrt(n)

# Calculate the Cohen's d or Z-score equivalent (how many standard errors fit into the mean)
z_stat = mean_improvement / standard_error

print(f"Standard Error of the Mean: {standard_error:.2f}")
print(f"Statistical Z-score (Effect size in standard errors): {z_stat:.2f}")
```

```python
# Formulate the conclusion based on the statistical threshold (Z > 2)
print("\n--- Hypothesis Testing Conclusion ---")
if z_stat > 2.0:
    print("Result: Reject the Null Hypothesis (H0).")
    print("Conclusion: The training program is statistically effective. Employee productivity significantly increased.")
else:
    print("Result: Fail to reject the Null Hypothesis (H0).")
    print("Conclusion: The training program is not effective. Observed differences could be due to pure random chance.")
```

Generate time series data for the year 2023. Demonstrate slicing for the following intervals:

January to March
April to June
July to September
October to December
Generate a time series data for a specific period and demonstrate how to slice this data for different intervals.





```python
# Import NumPy for data generation and Pandas for time series tracking
import numpy as np
import pandas as pd

# Create a sequence of daily dates for the entire year of 2023
date_index = pd.date_range(start='2023-01-01', end='2023-12-31', freq='D')

# Set a random seed and generate 365 random integers (e.g., daily sales metrics)
np.random.seed(42)
random_values = np.random.randint(100, 500, size=len(date_index))

# Combine the dates and values into a Pandas Series with dates as the index
time_series = pd.Series(random_values, index=date_index)

# Display the first few rows of our time series data
print("Initial Time Series Data (First 5 days):")
print(time_series.head())
```

```python
# Slice interval 1: January to March (Q1)
q1_data = time_series['2023-01':'2023-03']

# Slice interval 2: April to June (Q2)
q2_data = time_series['2023-04':'2023-06']

# Slice interval 3: July to September (Q3)
q3_data = time_series['2023-07':'2023-09']

# Slice interval 4: October to December (Q4)
q4_data = time_series['2023-10':'2023-12']

# Display summary statistics for each sliced interval to verify the operation
print(f"\nJanuary to March (Q1) count of days: {len(q1_data)}")
print(f"April to June (Q2) count of days: {len(q2_data)}")
print(f"July to September (Q3) count of days: {len(q3_data)}")
print(f"October to December (Q4) count of days: {len(q4_data)}")
```

🌟 Exercise 10 : Basic Visualization
Instructions
Use Matplotlib to visualize a simple dataset created with NumPy (e.g., a line graph of random numbers).

```python
# Import NumPy for calculations and Matplotlib for plotting
import numpy as np
import matplotlib.pyplot as plt

# Create an array of 20 sequential integers representing days (X-axis)
days = np.arange(1, 21)

# Set random seed for consistency
np.random.seed(24)

# Generate 20 random numbers representing stock values (Y-axis)
stock_values = np.random.randint(50, 150, size=20)

# Display the raw arrays to check what we are going to plot
print("Days (X-axis):", days)
print("Stock Values (Y-axis):", stock_values)
```

```python
# Create a line graph using the generated data
# 'marker="o"' adds dots on data points, 'linestyle="-."' changes the line pattern, and 'color="b"' makes it blue
plt.plot(days, stock_values, marker='o', linestyle='-.', color='b', label='Stock Trend')

# Add a label to the horizontal axis (X-axis)
plt.xlabel("Days of Observation")

# Add a label to the vertical axis (Y-axis)
plt.ylabel("Stock Value ($)")

# Add a title to the top of the graph
plt.title("Daily Stock Value Fluctuations")

# Enable the background grid lines for better readability
plt.grid(True)

# Add a legend based on the 'label' parameter in plt.plot
plt.legend()

# Render and display the final window with the graph
plt.show()
```
